```python
import pandas as pd

file_path = "total_data.csv"
df = pd.read_csv(file_path)

# '주차' 컬럼을 날짜 형식으로 변환
df['주차'] = pd.to_datetime(df['주차'], format='%Y%m%d')

# 같은 주차의 같은 소품종일 경우 그룹화 및 합산
grouped_df = df.groupby(['주차', '라인', '성별', '기획년도', '시즌이월', '상품년차', '시즌', '복종', '소품종']).sum(numeric_only=True).reset_index()

# 결과 확인
grouped_df.shape
# df.shape
```

```python
# 주어진 데이터의 최소/최대 날짜 범위 가져오기
date_range = pd.date_range(start=grouped_df['주차'].min(), end=grouped_df['주차'].max(), freq='W')
# date_range # 확인

# 누락된 주차 찾기
missing_weeks = sorted(set(date_range) - set(grouped_df['주차']))

# 누락된 주차 출력
print("누락된 주차:", missing_weeks)
```

```python
filtered_df = grouped_df[~grouped_df['라인'].isin(['ZD', 'ZE', 'ZF'])]

# grouped_df['소품종'].unique() #37개

# '복종'과 '소품종' 컬럼을 결합하여 새로운 '카테고리' 컬럼 생성
filtered_df['카테고리'] = filtered_df['복종'] + "_" + filtered_df['소품종'] +  "_" + filtered_df['라인'] 
print(sorted(filtered_df['카테고리'].unique()))
# filtered_df['카테고리'].nunique()
```

```python
filtered_df.shape
```

```python
# filtered_df['복종'].unique()
filtered2_df = filtered_df[~filtered_df['복종'].isin(['소품', '언더웨어'])]
filtered2_df['카테고리'].nunique()
print(sorted(filtered2_df['카테고리'].unique()))
```

```python
# filtered2_df.shape
filtered3_df = filtered2_df[['주차','카테고리','판매액','판매수량','매출원가','판매택가']]
filtered3_df
```

```python
filtered3_df['실판매가'] = filtered3_df['판매액']/filtered3_df['판매수량']

filtered3_df['택가'] = filtered3_df['판매택가']/filtered3_df['판매수량']

filtered3_df['할인율'] = (filtered3_df['택가'] - filtered3_df['실판매가']) / filtered3_df['택가']

filtered3_df.fillna(0, inplace=True)
filtered3_df
```
