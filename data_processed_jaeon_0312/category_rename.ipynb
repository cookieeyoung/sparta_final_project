## [기획년도, 주차, 카테고리_통합] 집계

```python
agg_dict = {
    '시즌이월': 'first', 
    '시즌': 'first', 
    '총입고수량': 'sum',
    '판매수량' : 'sum', 
    '판매액': 'sum', 
    '평균택가': 'mean', 
    '평균원가': 'mean',
    '주차별_평균_실판매가': 'mean',
    '월별_평균_실판매가': 'mean',
    '시즌별_평균_실판매가': 'mean'
}

grouped_df = model_df.groupby(['기획년도', '주차', '카테고리_통합']).agg(agg_dict).reset_index()
grouped_df.head()
```

```python
grouped_df.isna().sum()
```

```python
columns_to_round = ['평균택가','평균원가', '주차별_평균_실판매가', '월별_평균_실판매가', '시즌별_평균_실판매가']

grouped_df[columns_to_round] = grouped_df[columns_to_round].round(0)
grouped_df.head()
```

```python
grouped_df['총입고원가'] = grouped_df['총입고수량'] * grouped_df['평균원가']
grouped_df['총입고택가'] = grouped_df['총입고수량'] * grouped_df['평균택가']
grouped_df['매출원가계'] = grouped_df['판매수량'] * grouped_df['평균원가']
grouped_df['판매택가계'] = grouped_df['판매수량'] * grouped_df['평균택가']
grouped_df
```

```python
columns_to_round = ['평균택가','평균원가', '총입고원가', '총입고택가', '매출원가계', '판매택가계']

grouped_df[columns_to_round] = grouped_df[columns_to_round].round(0).astype('int64')
grouped_df.head()
```

```python
grouped_df['실판매가'] = grouped_df['판매액'] / grouped_df['판매수량']
grouped_df.head()
```

```python
grouped_df['실판매가'].apply(np.isinf).sum()
```

```python
grouped_df['실판매가'].replace([np.inf, -np.inf], np.nan, inplace=True)
grouped_df['실판매가'].apply(np.isinf).sum()
```

```python
grouped_df['실판매가'].isna().sum()
```

```python
grouped_df[grouped_df['실판매가'] <=0]
```

```python
grouped_df['실판매가'] = grouped_df['실판매가'].mask((grouped_df['실판매가'] <= 0) | (grouped_df['실판매가'].isna()), grouped_df['주차별_평균_실판매가'])

grouped_df['실판매가'] = grouped_df['실판매가'].fillna(grouped_df['월별_평균_실판매가'])

grouped_df['실판매가'] = grouped_df['실판매가'].fillna(grouped_df['시즌별_평균_실판매가'])

grouped_df.head()
```

```python
grouped_df[grouped_df['실판매가'] <= 0]
```

```python
grouped_df[grouped_df['실판매가'].isna()]
```

```python
grouped_df['할인율'] = ((grouped_df['평균택가'] - grouped_df['실판매가']) / grouped_df['평균택가']) * 100
grouped_df.head()
```

```python
columns_to_round = ['실판매가', '할인율']

grouped_df[columns_to_round] = grouped_df[columns_to_round].round(0).astype('int64')
grouped_df.head()
```

```python
grouped_df.dtypes
```

```python
grouped_df['누적판매수량'] = grouped_df.groupby(['기획년도', '카테고리_통합'])['판매수량'].cumsum()
grouped_df['누적판매액'] = grouped_df.groupby(['기획년도','카테고리_통합'])['판매액'].cumsum()
grouped_df['누적매출원가'] = grouped_df.groupby(['기획년도','카테고리_통합'])['매출원가계'].cumsum()
grouped_df['누적판매택가'] = grouped_df.groupby(['기획년도','카테고리_통합'])['판매택가계'].cumsum()
        
grouped_df['누적판매율'] = (grouped_df['누적판매수량'] / grouped_df['총입고수량']* 100).round(2)
grouped_df['ROI'] = ((grouped_df['누적판매액']/1.1 - grouped_df['누적매출원가'])/grouped_df['총입고원가']).round(2)
grouped_df.head()
```

```python
merged_df = grouped_df.merge(weather_df, on='주차', how='left')
```
